## 参考实现 `w04`（含依赖导入）

```python
import json
import random
import re
import sys
import numpy as np  # 通用科学计算


def w04():
    class Device:
        def __init__(self, dev_id, name, temp=25.0):
            self.dev_id, self.name, self.temp = dev_id, name, temp
            self.status = "停止"

        def start(self):
            if self.status == "运行":
                print(f"{self.name} 已在运行, 请勿重复启动")
                return
            self.status = "运行"
            print(f"{self.name} 启动成功")

        def stop(self):
            self.status = "停止"
            print(f"{self.name} 已停止")

        def get_status(self):
            return f"{self.dev_id} {self.name}: {self.status} {self.temp}℃"

        def __str__(self):
            return self.get_status()

    d1, d2 = Device("DEV-01", "空压机A", 41.2), Device("DEV-02", "冷却泵")
    d1.start()
    d1.start()
    d2.start()
    d1.stop()
    print(d1)
    print(d2.get_status())
```
