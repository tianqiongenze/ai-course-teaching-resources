### Starter 1：`# -*- coding: utf-8 -*-`

```python
# -*- coding: utf-8 -*-
"""M1.2 特征工程优化 —— 学生代码框架
先运行 m11 完成版获得基线, 在此逐步添加优化并记录消融表。
"""
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import GridSearchCV, cross_val_score, train_test_split

DATA = "data/pump_fault.csv"
ABLATION = []   # 消融记录: (步骤名, 准确率, F1)
```

### Starter 2：`def load_data(path):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def load_data(path):
    data = np.genfromtxt(path, delimiter=",", skip_header=1)
    return data[:, :5], data[:, 5].astype(int)
```

### Starter 3：`def evaluate(tag, Xtr, Xte, ytr, yte, model=No`
（Investigate: 读代码→Modify: 补全TODO）

```python
def evaluate(tag, Xtr, Xte, ytr, yte, model=None):
    model = model or RandomForestClassifier(n_estimators=100, random_state=42)
    model.fit(Xtr, ytr)
    acc = accuracy_score(yte, model.predict(Xte))
    f1 = f1_score(yte, model.predict(Xte), average="macro")
    ABLATION.append((tag, acc, f1))
    print(f"[消融] {tag:<22} 准确率={acc:.3f}  F1={f1:.3f}")
    return acc
```

### Starter 4：`def clean_outliers(X, y):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def clean_outliers(X, y):
    """清洗噪声样本: 对每个特征按3σ截断(记录删除条数)"""
    z = (X - X.mean(0)) / X.std(0)
    mask = (np.abs(z) < 3).all(axis=1)
    print(f"3σ清洗: {len(X)} → {int(mask.sum())} 条（删除 {int((~mask).sum())}）")
    return X[mask], y[mask]
```

### Starter 5：`def build_features(X):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def build_features(X):
    """派生特征≥2个: vibration/current, temp*pressure → (N, 5+k)"""
    rpm, temp, press, vib, amp = X.T
    return np.c_[X, vib / np.maximum(amp, 1e-6), temp * press]
```

### Starter 6：`def main():`
（Investigate: 读代码→Modify: 补全TODO）

```python
def main():
    X, y = load_data(DATA)
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25,
                                          stratify=y, random_state=42)
    base = evaluate("基线(原始)", Xtr, Xte, ytr, yte)

    # 步骤1: 清洗
    Xc, yc = clean_outliers(X, y)
    Xtr2, Xte2, ytr2, yte2 = train_test_split(Xc, yc, test_size=0.25,
                                              stratify=yc, random_state=42)
    evaluate("+清洗", Xtr2, Xte2, ytr2, yte2)

    # 【参考答案】TODO-3 +新特征: 同一规则(build_features)同时作用于训练/测试集
    Xf = build_features(Xc)
    Xtr3, Xte3, ytr3, yte3 = train_test_split(Xf, yc, test_size=0.25,
                                              stratify=yc, random_state=42)
    evaluate("+派生特征", Xtr3, Xte3, ytr3, yte3)

    # 【参考答案】TODO-4 +归一化: 树模型对单调缩放不敏感 → 实测后写进决策
    from sklearn.preprocessing import StandardScaler
    sc = StandardScaler().fit(Xtr3)
    evaluate("+派生特征+标准化", sc.transform(Xtr3), sc.transform(Xte3), ytr3, yte3)
    print("决策: 缩放不改变RF的分裂点排序, 本项无收益 → 不进生产方案(线性模型才需要缩放)。")

    # 【参考答案】TODO-5 +网格搜索: 只在训练集上拟合, 测试集只用于最终评估
    gs = GridSearchCV(RandomForestClassifier(random_state=42),
                      {"n_estimators": [50, 100, 200], "max_depth": [None, 8]}, cv=5)
    gs.fit(Xtr3, ytr3)
    pred = gs.predict(Xte3)
    acc = accuracy_score(yte3, pred)
    f1 = f1_score(yte3, pred, average="macro")
    ABLATION.append(("+网格搜索", acc, f1))
    print(f"[消融] +网格搜索                准确率={acc:.3f}  F1={f1:.3f}  最优参数={gs.best_params_}")

    # 【参考答案】TODO-6 5折交叉验证稳定性(用最优估计器, 在全量清洗+特征数据上)
    scores = cross_val_score(gs.best_estimator_, Xf, yc, cv=5)
    print(f"5折CV稳定性: 均值={scores.mean():.3f} ± {scores.std():.3f}  各折={np.round(scores, 3)}")
    print(f"生产推荐配置: {gs.best_params_} + 2个派生特征 + 3σ清洗 + 不做标准化"
          f"(折间波动{scores.std():.3f}, 波动<0.02 视为稳定)")

    print("\n===== 消融实验表 =====")
    for tag, acc, f1 in ABLATION:
        print(f"{tag:<24}{acc:>8.3f}{f1:>8.3f}")
    print(f"基线→最优提升: {(max(a for _, a, _ in ABLATION) - base) * 100:.1f}%")
```

## 教师参考答案（来源：`教师参考/solutions.py` · `m12()`）

> **发放纪律：本附录勿随学生版下发。**

上方学生版 TODO 与下方参考实现的对应关系：

| 学生版函数 | 说明 |
|---|---|
| `main()` | 见下方 `m12()` 内对应逻辑 |

```python
def m12():
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.metrics import accuracy_score, f1_score
    from sklearn.model_selection import (GridSearchCV, cross_val_score,
                                         train_test_split)

    data = np.genfromtxt("data/pump_fault.csv", delimiter=",", skip_header=1)
    X, y = data[:, :5], data[:, 5].astype(int)

    z = (X - X.mean(0)) / X.std(0)
    mask = (np.abs(z) < 3).all(axis=1)                # 3σ清洗
    Xc, yc = X[mask], y[mask]
    print(f"清洗: {len(X)} → {len(Xc)}")

    rpm, temp, press, vib, amp = Xc.T
    Xf = np.c_[Xc, vib / amp, temp * press]           # 派生特征×2

    def ev(tag, Xa):
        Xtr, Xte, ytr, yte = train_test_split(Xa, yc, test_size=0.25,
                                              stratify=yc, random_state=42)
        m = RandomForestClassifier(n_estimators=100, random_state=42).fit(Xtr, ytr)
        acc = accuracy_score(yte, m.predict(Xte))
        f1 = f1_score(yte, m.predict(Xte), average="macro")
        print(f"[消融] {tag:<16} acc={acc:.3f} F1={f1:.3f}")
        return acc

    b = ev("基线", Xc)
    ev("+派生特征", Xf)
    gs = GridSearchCV(RandomForestClassifier(random_state=42),
                      {"n_estimators": [50, 100, 200]}, cv=5)
    Xtr, Xte, ytr, yte = train_test_split(Xf, yc, test_size=0.25,
                                          stratify=yc, random_state=42)
    gs.fit(Xtr, ytr)
    a2 = accuracy_score(yte, gs.predict(Xte))
    print(f"[消融] +网格搜索      acc={a2:.3f} 最优参数={gs.best_params_}")
    cv = cross_val_score(gs.best_estimator_, Xf, yc, cv=5)
    print(f"5折CV: {cv.mean():.3f} ± {cv.std():.3f}")
    print(f"提升: {(a2 - b) * 100:.1f}%")
    print("决策: 树模型对归一化不敏感(矩阵中此项打低分), 故缩放不采纳。")
```
