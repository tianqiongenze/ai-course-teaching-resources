### Starter 2：`def check_env():`
（Investigate: 读代码→Modify: 补全TODO）

```python
def check_env():
    try:
        import sklearn  # noqa
        ok = True
    except ImportError:
        ok = False
    print("numpy:", np.__version__, "| sklearn:", "OK" if ok else "缺失!")
    print("环境OK" if ok else "请先安装scikit-learn")
```

### Starter 3：`def load_data(path):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def load_data(path):
    """读取CSV(仅用NumPy): 返回 X(float, N×5), y(int, N,)"""
    data = np.genfromtxt(path, delimiter=",", skip_header=1)
    return data[:, :5], data[:, 5].astype(int)
```

### Starter 4：`def split(X, y, test_ratio=0.25, seed=42):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def split(X, y, test_ratio=0.25, seed=42):
    """分层划分训练/测试集(每类按比例抽)"""
    rng = np.random.RandomState(seed)
    tr, te = [], []
    for c in np.unique(y):
        idx = np.where(y == c)[0]
        rng.shuffle(idx)
        k = int(len(idx) * (1 - test_ratio))
        tr.append(idx[:k])
        te.append(idx[k:])
    tr, te = np.concatenate(tr), np.concatenate(te)
    rng.shuffle(tr)
    rng.shuffle(te)
    return X[tr], X[te], y[tr], y[te]
```

### Starter 5：`def main():`
（Investigate: 读代码→Modify: 补全TODO）

```python
def main():
    X, y = load_data(DATA)
    print("数据形状:", X.shape, "类别分布:", np.bincount(y))

    from sklearn.linear_model import LogisticRegression
    from sklearn.tree import DecisionTreeClassifier
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.metrics import (accuracy_score, precision_score,
                                 recall_score, f1_score)

    Xtr, Xte, ytr, yte = split(X, y)

    models = {
        "逻辑回归": LogisticRegression(max_iter=1000),
        "决策树": DecisionTreeClassifier(max_depth=6, random_state=42),
        "随机森林": RandomForestClassifier(n_estimators=100, random_state=42),
    }
    print(f"{'模型':<8}{'准确率':>8}{'精确率':>8}{'召回率':>8}{'F1':>8}")
    best = (None, 0.0)
    for name, model in models.items():
        model.fit(Xtr, ytr)
        p = model.predict(Xte)
        acc = accuracy_score(yte, p)
        pr = precision_score(yte, p, average="macro")
        rc = recall_score(yte, p, average="macro")
        f1 = f1_score(yte, p, average="macro")
        print(f"{name:<8}{acc:>8.3f}{pr:>8.3f}{rc:>8.3f}{f1:>8.3f}")
        if acc > best[1]:
            best = (name, acc)

    pred2 = models[best[0]].predict(Xte[:2])
    print("\n预测 vs 真实(前2条):",
          [LABELS[i] for i in pred2], "vs", [LABELS[i] for i in yte[:2]])
    print(f"决策结论: 交付模型={best[0]}(acc={best[1]:.3f}); "
          "随机森林可给特征重要度, 逻辑回归可作产线侧轻量基线。")
```

## 教师参考答案（来源：`教师参考/solutions.py` · `m11()`）

上方学生版 TODO 与下方参考实现的对应关系：

```python
def m11():
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.linear_model import LogisticRegression
    from sklearn.metrics import (accuracy_score, f1_score, precision_score,
                                 recall_score)
    from sklearn.tree import DecisionTreeClassifier

    data = np.genfromtxt("data/pump_fault.csv", delimiter=",", skip_header=1)
    X, y = data[:, :5], data[:, 5].astype(int)

    idx = np.random.RandomState(42).permutation(len(X))
    X, y = X[idx], y[idx]
    n = int(len(X) * 0.75)
    Xtr, Xte, ytr, yte = X[:n], X[n:], y[:n], y[n:]

    models = {"逻辑回归": LogisticRegression(max_iter=1000),
              "决策树": DecisionTreeClassifier(max_depth=6, random_state=42),
              "随机森林": RandomForestClassifier(n_estimators=100,
                                             random_state=42)}
    print(f"{'模型':<6}{'准确':>7}{'精确':>7}{'召回':>7}{'F1':>7}")
    best = (None, 0)
    for name, m in models.items():
        m.fit(Xtr, ytr)
        p = m.predict(Xte)
        acc = accuracy_score(yte, p)
        row = [precision_score(yte, p, average="macro"),
               recall_score(yte, p, average="macro"),
               f1_score(yte, p, average="macro")]
        print(f"{name:<6}{acc:>7.3f}" + "".join(f"{v:>7.3f}" for v in row))
        if acc > best[1]:
            best = (name, acc)
    print("决策矩阵结论: 随机森林精度最优且可给出特征重要度, 选为交付模型;"
          "逻辑回归可作为产线侧轻量基线。最终:", best)
```
